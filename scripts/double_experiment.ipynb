# De Novo Lead-Binding Protein Design — RFD3 Double-Motif Pipeline

```text
RFD3  →  ProteinMPNN  →  RF3
```

Adapted for the **double-motif lead-sensing project**. Two copies of the Pb²⁺-Cys₃ motif from
PbrR691 / 5GPE are embedded in a single de novo scaffold.

Key differences from the single-motif pipeline:

- Input JSONs come from `02_prepare_rfd3_double_inputs.sh` (not `01_prepare_rfd3_inputs.sh`).
- Each JSON in `rfdiffusion_inputs/per_design/` contains one double-motif design.
- The input PDB is already embedded in the payload (`payload["input"]`); no separate context PDB lookup is needed.
- Both motif copies share the same open-space direction (outward vector).
- Chains: original motif on C/D, copied motif on X/Y (default).
- MPNN fixes residues from **both** motif copies.
- RF3 validation checks Cys3 conservation in both copies independently.

## How to run

1. Run `02_prepare_rfd3_double_inputs.sh` first.
2. Set `PREP_EXPERIMENT` below to the folder it wrote (e.g. `exp_doubles_02`).
3. Run cells top to bottom. GPU stages print the `bsub < ...` command — submit and wait before continuing.

## 1. Load prepared double-motif RFD3 inputs

Reads the per-design JSONs from  produced by
.

Each JSON already contains the path to the double-motif PDB (),
so no separate context PDB lookup is needed.


````python
def load_payload(path: Path) -> tuple[str, dict]:
    """Load one RFD3 input JSON and return its design name and payload."""
    with path.open() as handle:
        data = json.load(handle)
    if len(data) != 1:
        raise ValueError(f"Expected exactly one design per RFD3 input JSON: {path}")
    design_name = next(iter(data))
    return design_name, data[design_name]


def classify_contig_style(contig: str) -> str:
    """Classify whether an RFD3 contig is binder-style or monomeric-style."""
    if "/0" in contig:
        return "binder"
    return "monomeric"


def condition_label_from_name(design_name: str) -> str:
    """Build a stable condition label from the double-motif design name.

    Example input:
        5gpe_pb_motif_r5_ori5A_single_double_ori5A_rot180deg

    Extracts:
        ori5A_rot180deg
    """
    match = re.search(r"(ori\w+_rot\w+)", design_name)
    if match:
        return match.group(1)
    return design_name


def write_runtime_rfd3_json(
    design_name: str,
    payload: dict,
    condition: str,
    outdir: Path,
) -> Path:
    """Write a runtime RFD3 JSON. The input PDB path comes from the payload itself."""
    outdir.mkdir(parents=True, exist_ok=True)
    runtime_json = outdir / f"rfd3_input_{condition}.json"
    with runtime_json.open("w") as handle:
        json.dump({design_name: payload}, handle, indent=2)
    return runtime_json.resolve()


# Discover per-design JSONs produced by 02_prepare_rfd3_double_inputs.sh.
per_design_dir = PREP_ROOT / "rfdiffusion_inputs" / "per_design"
per_design_jsons = sorted(per_design_dir.glob("*.json"))

if not per_design_jsons:
    raise RuntimeError(
        f"No per-design JSONs found under {per_design_dir}.\n"
        "Run 02_prepare_rfd3_double_inputs.sh first."
    )

runtime_json_dir = configs_dir / "rfd3_runtime_inputs"
manifest_rows = []

for input_json in per_design_jsons:
    design_name, payload = load_payload(input_json)
    contig = payload["contig"]
    contig_style = classify_contig_style(contig)
    condition = condition_label_from_name(design_name)

    # The double-motif PDB is embedded in the payload.
    input_pdb = Path(payload["input"])
    if not input_pdb.exists():
        print(f"[WARNING] PDB not found: {input_pdb}")

    runtime_json = write_runtime_rfd3_json(
        design_name=design_name,
        payload=payload,
        condition=condition,
        outdir=runtime_json_dir,
    )

    print(f"[OK] {condition}")
    print(f"     contig = {contig}")
    print(f"     input  = {input_pdb.name}")

    manifest_rows.append(
        {
            "design_name": design_name,
            "condition": condition,
            "contig": contig,
            "contig_style": contig_style,
            "input_pdb": str(input_pdb),
            "runtime_input_json": str(runtime_json),
            "has_select_fixed_atoms": "select_fixed_atoms" in payload,
            "ori_token": str(payload.get("ori_token", "")),
        }
    )

rfd3_manifest = pd.DataFrame(manifest_rows)
manifest_csv = configs_dir / "rfd3_input_manifest.csv"
rfd3_manifest.to_csv(manifest_csv, index=False)

print(f"\nManifest: {manifest_csv}")
print(f"Total conditions: {len(rfd3_manifest)}")
rfd3_manifest[["condition", "contig"]].head(10)
````

## 2. RFD3 — generate scaffold backbones

````python
import textwrap

def write_lsf_array_script(
    cmds_path: Path,
    script_path: Path,
    job_name: str,
    queue: str,
    cores: int,
    mem: str,
    time_limit: str,
    group_size: int,
    gpu_spec: str | None = None,
    activate_lines: str = "",
    extra_exports: str = "",
) -> Path:
    """Write a generic LSF job-array script that executes commands from a command file."""
    if group_size < 1:
        raise ValueError("group_size must be >= 1")

    commands = [line for line in cmds_path.read_text().splitlines() if line.strip()]
    n_cmds = len(commands)
    if n_cmds == 0:
        raise RuntimeError(f"No commands in {cmds_path}")

    n_arrays = math.ceil(n_cmds / group_size)

    bsub_lines = [
        "#!/usr/bin/env bash",
        f"#BSUB -q {queue}",
        f"#BSUB -J {job_name}[1-{n_arrays}]",
        f"#BSUB -n {cores}",
    ]
    if gpu_spec:
        bsub_lines.append(f'#BSUB -gpu "{gpu_spec}"')
    bsub_lines.extend(
        [
            f"#BSUB -W {time_limit}",
            f'#BSUB -R "rusage[mem={mem}]"',
            '#BSUB -R "span[hosts=1]"',
            f"#BSUB -o {logs_dir}/%J_%I.out",
            f"#BSUB -e {logs_dir}/%J_%I.err",
        ]
    )

    body = f"""
set -euo pipefail

mkdir -p "{logs_dir}"

{activate_lines.strip()}

{extra_exports.strip()}

CMDS_FILE="{cmds_path}"
GROUP_SIZE={group_size}
JOB_INDEX="${{LSB_JOBINDEX:-1}}"
START=$(( (JOB_INDEX - 1) * GROUP_SIZE ))
END=$(( START + GROUP_SIZE ))

echo "Job index: $JOB_INDEX"
echo "Running command indices: $START to $((END - 1))"
echo "Commands file: $CMDS_FILE"

i=0
while IFS= read -r cmd; do
    if [ -z "$cmd" ]; then
        continue
    fi

    if [ "$i" -ge "$START" ] && [ "$i" -lt "$END" ]; then
        echo "----------------------------------------"
        echo "Running command index $i"
        echo "$cmd"
        eval "$cmd"
    fi

    i=$((i + 1))
done < "$CMDS_FILE"

echo "Done at $(date)"
"""

    script = "\n".join(bsub_lines) + "\n" + textwrap.dedent(body).lstrip()
    script_path.parent.mkdir(parents=True, exist_ok=True)
    script_path.write_text(script)
    return script_path


rfd3_cmds = cmds_dir / "rfd3.cmds"

with rfd3_cmds.open("w") as handle:
    for _, row in rfd3_manifest.iterrows():
        condition = row["condition"]
        out_dir = diffusion_out_dir / condition
        out_dir.mkdir(parents=True, exist_ok=True)

        cmd = (
            f'rfd3 design '
            f'out_dir="{out_dir}" '
            f'inputs="{row["runtime_input_json"]}" '
            f'ckpt_path="{RFD3_CKPT_PATH}" '
            f'diffusion_batch_size={DIFFUSION_BATCH_SIZE} '
            f'n_batches={N_BATCHES} '
            f'low_memory_mode=True '
            f'inference_sampler.step_scale=3 '
            f'inference_sampler.gamma_0=0.2'
        )
        handle.write(cmd + "\n")

activate_rfd3 = f"""
module load cuda/12.4
source "{CONDA_SH}"
conda activate "{RFD3_CONDA_ENV}"
"""

exports_rfd3 = """
export DISABLE_CUEQUIVARIANCE=true
export PYTORCH_CUDA_ALLOC_CONF=expandable_segments:True
"""

rfd3_script = write_lsf_array_script(
    cmds_path=rfd3_cmds,
    script_path=submit_dir / "rfd3.sh",
    job_name="rfd3_pb_double",
    queue=QUEUE,
    cores=4,
    mem="10GB",
    time_limit="2:00",
    group_size=1,
    gpu_spec="num=1:mode=exclusive_process",
    activate_lines=activate_rfd3,
    extra_exports=exports_rfd3,
)

print(f"Wrote {rfd3_cmds}")
print(f"Wrote {rfd3_script}")
print(f"Backbones per condition: {DIFFUSION_BATCH_SIZE * N_BATCHES}")
print("\nSubmit:")
print(f"  bsub < {rfd3_script}")
````

## 3. Process RFD3 outputs

Run this section after the RFD3 jobs finish.

````python
def matching_structure_for_rfd3_json(json_path: Path) -> Path | None:
    stem = json_path.stem
    candidates = [
        json_path.with_name(stem + ".cif.gz"),
        json_path.with_name(stem + ".cif"),
    ]
    for candidate in candidates:
        if candidate.exists():
            return candidate.resolve()
    return None


json_paths = sorted(diffusion_out_dir.glob("*/*.json"))
if not json_paths:
    raise SystemExit(f"No RFD3 JSON metrics found under {diffusion_out_dir}. Run RFD3 first.")

rows = []
all_keys = {"condition", "json_path", "structure_path"}

for jp in json_paths:
    with jp.open() as handle:
        data = json.load(handle)

    metrics = data.get("metrics", {})
    condition = jp.parent.name
    structure_path = matching_structure_for_rfd3_json(jp)

    row = {
        "condition": condition,
        "json_path": str(jp.resolve()),
        "structure_path": str(structure_path) if structure_path else "",
    }

    for key, value in metrics.items():
        if key in ("diffused_com", "fixed_com") and isinstance(value, (list, tuple)) and len(value) == 3:
            row[f"{key}_x"], row[f"{key}_y"], row[f"{key}_z"] = value
            all_keys.update({f"{key}_x", f"{key}_y", f"{key}_z"})
        else:
            row[key] = value
            all_keys.add(key)

    rows.append(row)

fieldnames = ["condition", "json_path", "structure_path"] + sorted(
    k for k in all_keys if k not in {"condition", "json_path", "structure_path"}
)

rfd3_metrics_csv = scores_dir / "rfd3_metrics_with_json_path.csv"
with rfd3_metrics_csv.open("w", newline="") as handle:
    writer = csv.DictWriter(handle, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

df_rfd3 = pd.read_csv(rfd3_metrics_csv)
print(f"Wrote {len(df_rfd3)} rows -> {rfd3_metrics_csv}")
df_rfd3.head()
````

````python
metrics = [
    "alanine_content", "glycine_content", "helix_fraction", "loop_fraction",
    "sheet_fraction", "max_ca_deviation", "n_chainbreaks",
    "n_clashing.interresidue_clashes_w_backbone",
    "n_clashing.interresidue_clashes_w_sidechain",
    "non_loop_fraction", "radius_of_gyration",
]
metrics = [m for m in metrics if m in df_rfd3.columns]
print("Plotting:", metrics)

if metrics:
    n_cols = 4
    n_rows = math.ceil(len(metrics) / n_cols)
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(4 * n_cols, 3 * n_rows))
    axes = np.array(axes).flatten()
    for ax, col in zip(axes, metrics):
        series = df_rfd3[col].dropna()
        if series.empty:
            ax.axis("off"); continue
        ax.hist(series, bins=30)
        ax.set_title(col)
        ax.set_ylabel("count")
    for ax in axes[len(metrics):]:
        ax.axis("off")
    plt.suptitle("RFD3 backbone metrics — Pb-Cys3 double-motif campaign", y=1.02)
    plt.tight_layout()
    plt.show()
````

## 4. ProteinMPNN — design sequences on scaffold chain A

ProteinMPNN redesigns only **chain A** (the generated scaffold).

Chains C, D (original motif) and X, Y (copied motif) are fixed context and must not be redesigned.
Fixed residues for ProteinMPNN are read from  in the RFD3 output JSON,
covering both Cys3 copies.


````python
df_mpnn_input = df_filt.copy() if len(df_filt) > 0 else df_rfd3.copy()
df_mpnn_input = df_mpnn_input[df_mpnn_input["structure_path"].astype(str).str.len() > 0].copy()

if df_mpnn_input.empty:
    raise SystemExit("No RFD3 structures available for MPNN.")

# Motif 1: original Cys3 from 5GPE (chains C, D).
MOTIF1_FIXED_KEYS = ["C78", "C82", "D113", "D122"]

# Motif 2: copied Cys3 (chains X, Y — same residue numbers, different chains).
# These keys come from the chain_map applied by the prepare script.
# Default chain map: C->X, D->Y, so residue numbers are identical.
MOTIF2_FIXED_KEYS = ["X78", "X82", "Y113", "Y122"]

ALL_MOTIF_FIXED_KEYS = MOTIF1_FIXED_KEYS + MOTIF2_FIXED_KEYS

mpnn_cmds = cmds_dir / "mpnn.cmds"
fixed_report_rows = []

with mpnn_cmds.open("w") as handle:
    for _, row in df_mpnn_input.iterrows():
        structure = Path(row["structure_path"])
        condition = row["condition"]
        backbone_name = structure.name.replace(".cif.gz", "").replace(".cif", "")
        this_out = mpnn_out_dir / condition / backbone_name
        this_out.mkdir(parents=True, exist_ok=True)

        json_path = Path(row["json_path"])
        if not json_path.exists():
            raise FileNotFoundError(f"Missing RFD3 JSON: {json_path}")

        with json_path.open() as f:
            rfd3_data = json.load(f)

        mapping = rfd3_data["diffused_index_map"]

        # Fix residues from both motif copies.
        # Keys present in mapping depend on what RFD3 tracked; skip missing ones with a warning.
        fixed_keys_found = []
        fixed_keys_missing = []
        for key in ALL_MOTIF_FIXED_KEYS:
            if key in mapping:
                fixed_keys_found.append(key)
            else:
                fixed_keys_missing.append(key)

        if fixed_keys_missing:
            print(f"[WARNING] {backbone_name}: keys not in mapping: {fixed_keys_missing}")

        if not fixed_keys_found:
            raise KeyError(f"No motif keys found in diffused_index_map for {json_path}")

        # ProteinMPNN-style fixed residue strings are normally space-separated.
        # Build fixed residue IDs from the RFD3 residue mapping.
        fixed_residue_ids = [mapping[key] for key in fixed_keys_found]
        fixed_residues_report = ",".join(fixed_residue_ids)

        config_path = this_out / f"{backbone_name}_mpnn_config.json"

        mpnn_config = {
            "model_type": "protein_mpnn",
            "checkpoint_path": str(PROTEINMPNN_CKPT),
            "is_legacy_weights": True,
            "out_directory": str(this_out),
            "seed": MPNN_SEED,
            "batch_size": MPNN_BATCH_SIZE,
            "number_of_batches": MPNN_NUMBER_OF_BATCHES,
            "inputs": [
                {
                    "structure_path": str(structure),
                    "name": backbone_name,
                    "fixed_residues": fixed_residue_ids,
                }
            ],
        }

        with config_path.open("w") as f:
            json.dump(mpnn_config, f, indent=2)

        fixed_report_rows.append(
            {
                "condition": condition,
                "backbone_name": backbone_name,
                "structure_path": str(structure),
                "json_path": str(json_path),
                "config_path": str(config_path),
                "fixed_keys": ",".join(fixed_keys_found),
                "missing_keys": ",".join(fixed_keys_missing),
                "fixed_residues": fixed_residues_report,
            }
        )

        cmd = f'mpnn --config_json "{config_path}"'
        handle.write(cmd + "\n")

df_fixed_report = pd.DataFrame(fixed_report_rows)
fixed_report_path = scores_dir / "mpnn_fixed_residues_report.csv"
df_fixed_report.to_csv(fixed_report_path, index=False)

activate_mpnn = f"""
source "{CONDA_SH}"
conda activate "{MPNN_CONDA_ENV}"
"""

exports_mpnn = """
export DISABLE_CUEQUIVARIANCE=true
export PYTORCH_CUDA_ALLOC_CONF=expandable_segments:True
"""

mpnn_script = write_lsf_array_script(
    cmds_path=mpnn_cmds,
    script_path=submit_dir / "mpnn.sh",
    job_name="mpnn_pb_double_fixed_cys3",
    queue=QUEUE,
    cores=4,
    mem="10GB",
    time_limit="2:00",
    group_size=10,
    gpu_spec=None,
    activate_lines=activate_mpnn,
    extra_exports=exports_mpnn,
)

print(f"Wrote {mpnn_cmds}")
print(f"Wrote {mpnn_script}")
print(f"Fixed report: {fixed_report_path}")
print(f"Backbones going to MPNN: {len(df_mpnn_input)}")
print(f"Sequences per backbone : {MPNN_BATCH_SIZE * MPNN_NUMBER_OF_BATCHES}")
print("\nFixed residues preview:")
display(df_fixed_report.head(20))
print("\nSubmit:")
print(f"  bsub < {mpnn_script}")
````

## 5. RF3 — fold scaffold + SpyTag with Pb-Cys₃ context

RF3 folds:



For double-motif designs, RF3 receives only the scaffold sequence (chain A).
The fixed Pb-Cys₃ context (chains C/D and X/Y) is not passed as a separate chain to RF3 —
validation is done post-hoc via RMSD against the preparation-stage double-motif PDB.


````python
rf3_json_dir = Path(configs_dir) / "rf3"
rf3_json_dir.mkdir(parents=True, exist_ok=True)


def make_monomer_rf3_json(name: str, seq: str) -> dict:
    return {
        "name": name,
        "components": [{"chain_id": "A", "seq": seq}],
    }


DESIGN_RE = re.compile(r"_b(\d+)_d(\d+)")


def parse_fasta(path):
    header, chunks = None, []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if header is not None:
                    yield header, "".join(chunks)
                header, chunks = line[1:], []
            else:
                chunks.append(line)
        if header is not None:
            yield header, "".join(chunks)


fa_files = sorted(glob.glob(os.path.join(mpnn_out_dir, "**", "*.fa"), recursive=True))
if not fa_files:
    raise SystemExit(f"No MPNN FASTAs under {mpnn_out_dir} — run MPNN first.")

written = 0
rf3_manifest_rows = []

for fa in fa_files:
    model_dir = os.path.basename(os.path.dirname(fa))
    for header, seq in parse_fasta(fa):
        m = DESIGN_RE.search(header)
        if not m:
            continue
        batch_idx = int(m.group(1))
        design_idx = int(m.group(2))
        chain_a = seq.split(":", 1)[0]
        name = f"{model_dir}__b{batch_idx}_d{design_idx}"
        j = make_monomer_rf3_json(name=name, seq=chain_a + SPYTAG)
        out_json = rf3_json_dir / f"{name}.json"
        with open(out_json, "w") as out:
            json.dump(j, out, indent=2)
        rf3_manifest_rows.append({
            "name": name,
            "rf3_json": str(out_json.resolve()),
            "mpnn_fasta": str(Path(fa).resolve()),
            "sequence": chain_a,
            "sequence_with_spytag": chain_a + SPYTAG,
            "sequence_length": len(chain_a),
        })
        written += 1

rf3_manifest = pd.DataFrame(rf3_manifest_rows)
rf3_manifest_path = configs_dir / "rf3_manifest.csv"
rf3_manifest.to_csv(rf3_manifest_path, index=False)

print(f"Wrote {written} RF3 input JSONs -> {rf3_json_dir}")
print(f"RF3 manifest: {rf3_manifest_path}")
rf3_manifest.head()
````

````python
from pathlib import Path
import json
import re
import math

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import biotite.structure as struc
from biotite.structure.io import load_structure as bio_load


# ============================================================
# 0. Paths and settings
# ============================================================

PROJECT_ROOT = Path("/dtu/blackhole/00/222976/lead_sensoring")

EXP_DESIGN_DIR = PROJECT_ROOT / "work/exp_doubles_01"
EXP_RUN_DIR = PROJECT_ROOT / "work/exp_doubles_01_rfd3_mpnn_rf3"

DOUBLE_MOTIF_REFERENCE_DIR = EXP_DESIGN_DIR / "double_motif_structures"

try:
    DIFFUSION_OUT_DIR = Path(diffusion_out_dir)
except NameError:
    DIFFUSION_OUT_DIR = EXP_RUN_DIR / "diffusion_out"

SCORES_DIR = EXP_RUN_DIR / "scores"
PLOTS_DIR = EXP_RUN_DIR / "plots"

SCORES_DIR.mkdir(parents=True, exist_ok=True)
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

OUT_RMSD = SCORES_DIR / "rf3_double_motif_vs_original_motif_rmsd.csv"
OUT_RANKED = SCORES_DIR / "rf3_double_motif_ranked_by_motif_rmsd.csv"
OUT_PLOT = PLOTS_DIR / "rf3_ranked_worst_motif_rmsd.png"

# Pb-binding Cys3 only.
MOTIF1_CYS3_KEYS = ["C78", "D113", "D122"]
MOTIF2_CYS3_KEYS = ["X78", "Y113", "Y122"]

BACKBONE_ATOMS = ["N", "CA", "C", "O"]
SG_ATOMS = ["SG"]

PTM_CUT = 0.80
CYS3_BACKBONE_RMSD_CUT = 1.00
CYS3_SG_RMSD_CUT = 1.00

STRUCTURE_SUFFIXES = (".cif", ".pdb", ".cif.gz", ".pdb.gz")


# ============================================================
# 1. Display helper
# ============================================================

def safe_display(df, n=5):
    """Display a DataFrame in notebooks, otherwise print it."""
    try:
        display(df.head(n))
    except NameError:
        print(df.head(n).to_string())


# ============================================================
# 2. Basic structure helpers
# ============================================================

def load_atom_array(path):
    """Load a PDB/mmCIF file as a Biotite AtomArray."""
    path = Path(path)

    if not path.is_file():
        raise FileNotFoundError(f"Structure file not found: {path}")

    if str(path).endswith(".gz"):
        import gzip

        tmp = Path("/tmp/_tmp_biotite_load")
        if str(path).endswith(".cif.gz"):
            tmp = tmp.with_suffix(".cif")
        elif str(path).endswith(".pdb.gz"):
            tmp = tmp.with_suffix(".pdb")
        else:
            raise ValueError(f"Unsupported compressed structure format: {path}")

        with gzip.open(path, "rt") as f:
            tmp.write_text(f.read())

        arr = bio_load(str(tmp))
    else:
        arr = bio_load(str(path))

    if isinstance(arr, struc.AtomArrayStack):
        arr = arr[0]

    return arr


def parse_chain_res(label):
    """Parse residue labels like A42, C78, X113."""
    m = re.match(r"^([A-Za-z]+)(\d+)([A-Za-z]?)$", str(label))
    if not m:
        raise ValueError(f"Cannot parse residue label: {label}")
    return m.group(1), int(m.group(2))


def get_atom_coord(arr, chain, res_id, atom_name):
    """Return one atom coordinate from an AtomArray."""
    mask = (
        (arr.chain_id.astype(str) == str(chain))
        & (arr.res_id == int(res_id))
        & (arr.atom_name.astype(str) == str(atom_name))
    )

    if mask.sum() == 0:
        raise KeyError(f"{chain}{res_id}@{atom_name} not found")

    return arr.coord[mask][0]


def resolve_chain(atom_array, preferred_chain):
    """
    Use preferred_chain if present.

    If the structure has only one chain, use that single chain.
    This handles RF3 outputs that collapse everything to chain A.
    """
    chains = sorted(set(str(c) for c in atom_array.chain_id))

    if str(preferred_chain) in chains:
        return str(preferred_chain)

    if len(chains) == 1:
        return chains[0]

    raise KeyError(
        f"Chain {preferred_chain} not found. Available chains: {chains}"
    )


def kabsch_rmsd(P, Q):
    """Compute standard Kabsch RMSD between paired coordinates."""
    P = np.asarray(P, dtype=float)
    Q = np.asarray(Q, dtype=float)

    if P.shape != Q.shape:
        raise ValueError(f"Shape mismatch: P{P.shape} vs Q{Q.shape}")

    if len(P) < 3:
        return np.nan

    P_cent = P - P.mean(axis=0)
    Q_cent = Q - Q.mean(axis=0)

    H = P_cent.T @ Q_cent
    U, _, Vt = np.linalg.svd(H)

    d = np.linalg.det(Vt.T @ U.T)
    D = np.diag([1.0, 1.0, d])

    R = Vt.T @ D @ U.T
    P_rot = P_cent @ R.T

    return float(np.sqrt(np.mean(np.sum((P_rot - Q_cent) ** 2, axis=1))))


def rmsd_to_score(x, scale):
    """Convert RMSD to a soft score in [0, 1]."""
    if not np.isfinite(x):
        return np.nan
    return float(np.exp(-0.5 * (float(x) / float(scale)) ** 2))


def safe_nanmax(values):
    """Return nanmax without warning when all values are NaN."""
    values = [v for v in values if np.isfinite(v)]
    if not values:
        return np.nan
    return float(np.max(values))


# ============================================================
# 3. Strict ID helpers
# ============================================================

def remove_structure_suffix(name):
    """Remove only file suffixes, not semantic suffixes like _model_0."""
    s = Path(str(name)).name

    for suffix in [".cif.gz", ".pdb.gz", ".cif", ".pdb", ".json", ".csv"]:
        if s.endswith(suffix):
            s = s[: -len(suffix)]

    return s


def clean_export_prefix(name):
    """Remove manual export prefixes that should not affect identity."""
    s = str(name)
    s = re.sub(r"^BEST_TRUE_RF3_", "", s)
    return s


def infer_design_id_from_structure_path(path):
    """Infer a design ID from an RF3 output structure path."""
    path = Path(path)
    stem = remove_structure_suffix(path.name)

    generic_names = {
        "model",
        "model_0",
        "prediction",
        "structure",
        "output",
        "fold",
        "rank_0",
    }

    if stem in generic_names or len(stem) < 12:
        stem = remove_structure_suffix(path.parent.name)

    return clean_export_prefix(stem)


def rf3_design_to_rfd3_parent_id(design_id):
    """
    Convert an RF3 design ID into the exact parent RFD3 backbone ID.

    Example:
    rfd3_input_..._3_model_0__b0_d0
    -> rfd3_input_..._3_model_0

    Critical:
    Do not remove _model_0. It is part of the unique RFD3 identity.
    """
    s = remove_structure_suffix(design_id)
    s = clean_export_prefix(s)

    s = re.sub(r"__b\d+_d\d+.*$", "", s)

    return s


def ori_tag_from_text(text):
    """Extract ori13A, ori15A, ori18A style tag."""
    m = re.search(r"ori\d+A", str(text))
    return m.group(0) if m else None


# ============================================================
# 4. Build strict RFD3 records
# ============================================================

def build_rfd3_records(diffusion_out_dir):
    """
    Build RFD3 mapping records using only exact primary IDs.

    This deliberately avoids short aliases because they are ambiguous.
    """
    diffusion_out_dir = Path(diffusion_out_dir)

    json_files = sorted(diffusion_out_dir.rglob("*.json"))

    if not json_files:
        raise FileNotFoundError(
            f"No RFD3 JSON files found under {diffusion_out_dir}"
        )

    primary_records = {}

    for jp in json_files:
        try:
            with jp.open() as f:
                data = json.load(f)
        except Exception:
            continue

        mapping = data.get("diffused_index_map", {})

        if not mapping:
            continue

        primary_id = clean_export_prefix(remove_structure_suffix(jp.name))
        condition = jp.parent.name

        if primary_id in primary_records:
            old = primary_records[primary_id]["json_path"]
            raise ValueError(
                "Duplicate RFD3 primary ID found.\n"
                f"ID: {primary_id}\n"
                f"Old: {old}\n"
                f"New: {jp}"
            )

        primary_records[primary_id] = {
            "primary_id": primary_id,
            "condition": condition,
            "mapping": mapping,
            "json_path": str(jp),
        }

    print("RFD3 primary records:", len(primary_records))

    if not primary_records:
        raise ValueError(
            f"No JSON files with diffused_index_map were found in {diffusion_out_dir}"
        )

    return primary_records


def resolve_rfd3_record_from_rf3_design(design_id, rfd3_primary_records):
    """Resolve RF3 design ID to exact parent RFD3 record."""
    parent_id = rf3_design_to_rfd3_parent_id(design_id)

    if parent_id in rfd3_primary_records:
        return parent_id, rfd3_primary_records[parent_id]

    nearby = [
        k for k in rfd3_primary_records
        if parent_id in k or k in parent_id
    ]

    raise KeyError(
        "Could not resolve exact RFD3 parent.\n"
        f"RF3 design ID: {design_id}\n"
        f"Expected parent ID: {parent_id}\n"
        f"Nearby candidates: {nearby[:10]}\n"
        f"First RFD3 IDs: {list(rfd3_primary_records.keys())[:10]}"
    )


rfd3_primary_records = build_rfd3_records(DIFFUSION_OUT_DIR)


# ============================================================
# 5. Collect RF3 structures
# ============================================================

def find_rf3_structure_files(run_dir):
    """
    Find RF3 output structures.

    We keep only files/folders containing the RF3 batch/design suffix:
    __b0_d0, __b1_d2, etc.
    """
    run_dir = Path(run_dir)

    all_files = [
        p for p in run_dir.rglob("*")
        if p.is_file() and str(p).endswith(STRUCTURE_SUFFIXES)
    ]

    skip_parts = {
        "diffusion_out",
        "double_motif_structures",
        "motif_scan",
        "rfdiffusion_inputs",
        "best_design_export",
        "plots",
        "scores",
        "submit",
        "logs",
    }

    filtered = []

    for p in all_files:
        lower_parts = {part.lower() for part in p.parts}

        if lower_parts.intersection(skip_parts):
            continue

        path_text = str(p)
        design_id = infer_design_id_from_structure_path(p)

        has_rf3_suffix = (
            re.search(r"__b\d+_d\d+", path_text) is not None
            or re.search(r"__b\d+_d\d+", design_id) is not None
        )

        if has_rf3_suffix:
            filtered.append(p)

    filtered = sorted(set(filtered))

    if not filtered:
        raise FileNotFoundError(
            "No RF3 structures with __bX_dY suffix were found. "
            "Your RF3 output folder naming may be different."
        )

    return filtered


def recursive_find_key(obj, wanted_keys):
    """Recursively find numeric values for selected keys in JSON-like data."""
    hits = []

    if isinstance(obj, dict):
        for k, v in obj.items():
            if str(k).lower() in wanted_keys:
                try:
                    hits.append(float(v))
                except Exception:
                    pass
            hits.extend(recursive_find_key(v, wanted_keys))

    elif isinstance(obj, list):
        for x in obj:
            hits.extend(recursive_find_key(x, wanted_keys))

    return hits


def find_ptm_near_structure(structure_path):
    """Try to find pTM in JSON files close to the RF3 structure."""
    structure_path = Path(structure_path)

    candidate_jsons = []
    candidate_jsons.extend(structure_path.parent.glob("*.json"))
    candidate_jsons.extend(structure_path.parent.parent.glob("*.json"))

    wanted = {"ptm", "predicted_tm_score", "tm_score"}

    for jp in sorted(set(candidate_jsons)):
        try:
            with jp.open() as f:
                data = json.load(f)

            hits = recursive_find_key(data, wanted)
            if hits:
                return float(hits[0]), str(jp)

        except Exception:
            continue

    return np.nan, ""


def build_rf3_output_table(run_dir):
    """Build a clean RF3 output table from structures on disk."""
    structure_files = find_rf3_structure_files(run_dir)

    rows = []

    for p in structure_files:
        design_id = infer_design_id_from_structure_path(p)
        parent_rfd3_id = rf3_design_to_rfd3_parent_id(design_id)
        ptm, ptm_source = find_ptm_near_structure(p)

        rows.append(
            {
                "design_id": design_id,
                "rfd3_parent_id": parent_rfd3_id,
                "rf3_structure_path": str(p),
                "ptm": ptm,
                "ptm_source": ptm_source,
            }
        )

    df = pd.DataFrame(rows)
    df = df.drop_duplicates(subset=["rf3_structure_path"]).reset_index(drop=True)

    print("RF3 structures found:", len(df))
    safe_display(df, 8)

    return df


df_rf3_outputs = build_rf3_output_table(EXP_RUN_DIR)


# ============================================================
# 6. Reference structure helper
# ============================================================

def get_reference_path_for_design(parent_id, rec):
    """
    Find the original double-motif reference PDB for this design.

    This assumes one reference PDB per ori condition.
    """
    pdb_files = sorted(DOUBLE_MOTIF_REFERENCE_DIR.glob("*.pdb"))

    if not pdb_files:
        raise FileNotFoundError(
            f"No reference PDB files found in {DOUBLE_MOTIF_REFERENCE_DIR}"
        )

    search_text = " ".join(
        str(x)
        for x in [
            parent_id,
            rec.get("primary_id", ""),
            rec.get("condition", ""),
            rec.get("json_path", ""),
        ]
    )

    ori_tag = ori_tag_from_text(search_text)

    if ori_tag is None:
        raise ValueError(f"Could not infer ori tag from: {search_text}")

    matches = [p for p in pdb_files if ori_tag in p.stem]

    if len(matches) != 1:
        raise ValueError(
            f"Expected one reference for {ori_tag}, found {len(matches)}: "
            f"{[p.name for p in matches]}"
        )

    return matches[0]


# ============================================================
# 7. Motif RMSD calculation
# ============================================================

def collect_motif_rmsd(rf3_arr, ref_arr, mapping, motif_keys, atoms):
    """
    Compare RF3 final motif to the original double-motif reference.

    RF3 side:
        mapping[key], e.g. C78 -> A48

    Reference side:
        key itself, e.g. C78 in the original double-motif PDB.
    """
    P = []
    Q = []
    used_atoms = []
    missing_atoms = []

    for key in motif_keys:
        if key not in mapping:
            missing_atoms.append(f"{key}:missing_from_mapping")
            continue

        rf3_label = str(mapping[key])
        rf3_chain_raw, rf3_res = parse_chain_res(rf3_label)

        ref_chain_raw, ref_res = parse_chain_res(key)

        try:
            rf3_chain = resolve_chain(rf3_arr, rf3_chain_raw)
            ref_chain = resolve_chain(ref_arr, ref_chain_raw)
        except Exception as exc:
            missing_atoms.append(f"{key}->{rf3_label}:chain_error:{exc}")
            continue

        for atom in atoms:
            try:
                rf3_coord = get_atom_coord(rf3_arr, rf3_chain, rf3_res, atom)
                ref_coord = get_atom_coord(ref_arr, ref_chain, ref_res, atom)

                P.append(rf3_coord)
                Q.append(ref_coord)
                used_atoms.append(f"{key}->{rf3_label}@{atom}")

            except Exception:
                missing_atoms.append(f"{key}->{rf3_label}@{atom}")

    if len(P) < 3:
        return np.nan, len(P), ",".join(used_atoms), ",".join(missing_atoms)

    return (
        kabsch_rmsd(np.asarray(P), np.asarray(Q)),
        len(P),
        ",".join(used_atoms),
        ",".join(missing_atoms),
    )


def check_cys_preserved(rf3_arr, mapping, motif_keys):
    """Check whether RF3 kept the mapped motif residues as CYS."""
    checked = []
    failed = []

    for key in motif_keys:
        if key not in mapping:
            failed.append(f"{key}:missing_from_mapping")
            return False, ",".join(checked), ",".join(failed)

        rf3_label = str(mapping[key])
        chain_raw, res_id = parse_chain_res(rf3_label)

        try:
            chain = resolve_chain(rf3_arr, chain_raw)
        except Exception as exc:
            failed.append(f"{key}->{rf3_label}:chain_error:{exc}")
            return False, ",".join(checked), ",".join(failed)

        mask = (
            (rf3_arr.chain_id.astype(str) == str(chain))
            & (rf3_arr.res_id == int(res_id))
        )

        if mask.sum() == 0:
            failed.append(f"{key}->{rf3_label}:residue_not_found")
            return False, ",".join(checked), ",".join(failed)

        res_name = str(rf3_arr.res_name[mask][0])
        has_sg = bool(np.any(rf3_arr.atom_name[mask].astype(str) == "SG"))

        checked.append(f"{key}->{rf3_label}:{res_name}:has_SG={has_sg}")

        if res_name != "CYS":
            failed.append(f"{key}->{rf3_label}:expected_CYS_found_{res_name}")
            return False, ",".join(checked), ",".join(failed)

        if not has_sg:
            failed.append(f"{key}->{rf3_label}:CYS_without_SG")
            return False, ",".join(checked), ",".join(failed)

    return True, ",".join(checked), ",".join(failed)


# ============================================================
# 8. Evaluate RF3 structures
# ============================================================

rmsd_rows = []
failed_rows = []

for _, row in df_rf3_outputs.iterrows():
    design_id = row["design_id"]
    parent_id = row["rfd3_parent_id"]
    rf3_path = Path(row["rf3_structure_path"])

    try:
        resolved_parent_id, rec = resolve_rfd3_record_from_rf3_design(
            design_id,
            rfd3_primary_records,
        )

        mapping = rec["mapping"]
        ref_path = get_reference_path_for_design(resolved_parent_id, rec)

        rf3_arr = load_atom_array(rf3_path)
        ref_arr = load_atom_array(ref_path)

        kept1, cys_checked1, cys_failed1 = check_cys_preserved(
            rf3_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
        )

        kept2, cys_checked2, cys_failed2 = check_cys_preserved(
            rf3_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
        )

        bb1, bb_n1, bb_atoms1, bb_missing1 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
            BACKBONE_ATOMS,
        )

        bb2, bb_n2, bb_atoms2, bb_missing2 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
            BACKBONE_ATOMS,
        )

        sg1, sg_n1, sg_atoms1, sg_missing1 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
            SG_ATOMS,
        )

        sg2, sg_n2, sg_atoms2, sg_missing2 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
            SG_ATOMS,
        )

        ptm = float(row.get("ptm", np.nan))

        max_bb_rmsd = safe_nanmax([bb1, bb2])
        max_sg_rmsd = safe_nanmax([sg1, sg2])

        passes_first_triage = (
            kept1
            and kept2
            and np.isfinite(bb1)
            and np.isfinite(bb2)
            and np.isfinite(sg1)
            and np.isfinite(sg2)
            and bb1 <= CYS3_BACKBONE_RMSD_CUT
            and bb2 <= CYS3_BACKBONE_RMSD_CUT
            and sg1 <= CYS3_SG_RMSD_CUT
            and sg2 <= CYS3_SG_RMSD_CUT
            and (not np.isfinite(ptm) or ptm >= PTM_CUT)
        )

        rmsd_rows.append(
            {
                "design_id": design_id,
                "rfd3_parent_id": parent_id,
                "resolved_rfd3_parent_id": resolved_parent_id,
                "rfd3_primary_id": rec["primary_id"],
                "condition": rec["condition"],

                "rf3_structure_path": str(rf3_path),
                "original_motif_reference_path": str(ref_path),
                "rfd3_json_path": rec["json_path"],

                "ptm": ptm,
                "ptm_source": row.get("ptm_source", ""),

                "kept_cys3_motif1": kept1,
                "kept_cys3_motif2": kept2,
                "kept_both_cys3": kept1 and kept2,

                "cys_checked_motif1": cys_checked1,
                "cys_checked_motif2": cys_checked2,
                "cys_failed_motif1": cys_failed1,
                "cys_failed_motif2": cys_failed2,

                "cys3_backbone_rmsd_motif1": bb1,
                "cys3_backbone_rmsd_motif2": bb2,
                "cys3_sg_rmsd_motif1": sg1,
                "cys3_sg_rmsd_motif2": sg2,

                "max_cys3_backbone_rmsd": max_bb_rmsd,
                "max_cys3_sg_rmsd": max_sg_rmsd,

                "n_backbone_atoms_motif1": bb_n1,
                "n_backbone_atoms_motif2": bb_n2,
                "n_sg_atoms_motif1": sg_n1,
                "n_sg_atoms_motif2": sg_n2,

                "missing_backbone_atoms_motif1": bb_missing1,
                "missing_backbone_atoms_motif2": bb_missing2,
                "missing_sg_atoms_motif1": sg_missing1,
                "missing_sg_atoms_motif2": sg_missing2,

                "cys3_backbone_score_motif1": rmsd_to_score(
                    bb1,
                    CYS3_BACKBONE_RMSD_CUT,
                ),
                "cys3_backbone_score_motif2": rmsd_to_score(
                    bb2,
                    CYS3_BACKBONE_RMSD_CUT,
                ),
                "cys3_sg_score_motif1": rmsd_to_score(
                    sg1,
                    CYS3_SG_RMSD_CUT,
                ),
                "cys3_sg_score_motif2": rmsd_to_score(
                    sg2,
                    CYS3_SG_RMSD_CUT,
                ),

                "passes_first_triage": passes_first_triage,
            }
        )

    except Exception as exc:
        failed_rows.append(
            {
                "design_id": design_id,
                "rfd3_parent_id": parent_id,
                "rf3_structure_path": str(rf3_path),
                "error": str(exc),
            }
        )
        print("[FAILED]", design_id)
        print(exc)
        print()


df_rmsd = pd.DataFrame(rmsd_rows)
df_failed = pd.DataFrame(failed_rows)

df_rmsd.to_csv(OUT_RMSD, index=False)

print("Evaluated RF3 structures:", len(df_rmsd))
print("Failed RF3 structures:", len(df_failed))

if len(df_rmsd):
    print("Kept both Cys3 motifs:", int(df_rmsd["kept_both_cys3"].sum()))
    print("Passing first triage:", int(df_rmsd["passes_first_triage"].sum()))

print("Saved RMSD table:", OUT_RMSD)

safe_display(df_rmsd, 8)

if len(df_failed):
    print("\nFailed cases:")
    safe_display(df_failed, 10)


# ============================================================
# 9. Rank designs
# ============================================================

if len(df_rmsd):
    df_ranked = df_rmsd.copy()

    # Sort good designs first, then by worst motif RMSD.
    df_ranked = df_ranked.sort_values(
        by=[
            "passes_first_triage",
            "kept_both_cys3",
            "max_cys3_sg_rmsd",
            "max_cys3_backbone_rmsd",
            "ptm",
        ],
        ascending=[
            False,
            False,
            True,
            True,
            False,
        ],
        na_position="last",
    ).reset_index(drop=True)

    df_ranked.to_csv(OUT_RANKED, index=False)

    print("Saved ranked table:", OUT_RANKED)

    cols_to_show = [
        "design_id",
        "condition",
        "ptm",
        "kept_both_cys3",
        "max_cys3_backbone_rmsd",
        "max_cys3_sg_rmsd",
        "passes_first_triage",
        "rf3_structure_path",
    ]

    safe_display(df_ranked[cols_to_show], 12)

else:
    df_ranked = pd.DataFrame()
    print("No RMSD rows available. Ranking was not created.")


# ============================================================
# 10. Diagnostic summary for Cys failures
# ============================================================

if len(df_rmsd):
    failed_cys = df_rmsd[~df_rmsd["kept_both_cys3"]].copy()

    print("\nCys3 preservation summary:")
    print(df_rmsd["kept_both_cys3"].value_counts(dropna=False).to_string())

    if len(failed_cys):
        diagnostic_cols = [
            "design_id",
            "rfd3_parent_id",
            "kept_cys3_motif1",
            "kept_cys3_motif2",
            "cys_checked_motif1",
            "cys_checked_motif2",
            "cys_failed_motif1",
            "cys_failed_motif2",
            "missing_sg_atoms_motif1",
            "missing_sg_atoms_motif2",
        ]

        print("\nFirst Cys3 failure diagnostics:")
        safe_display(failed_cys[diagnostic_cols], 12)


# ============================================================
# 11. Plot ranked worst motif RMSD
# ============================================================

if len(df_ranked):
    plot_df = df_ranked.copy()
    plot_df["rank"] = np.arange(1, len(plot_df) + 1)

    fig, ax = plt.subplots(figsize=(9, 5))

    ax.scatter(
        plot_df["rank"],
        plot_df["max_cys3_backbone_rmsd"],
        s=35,
        alpha=0.85,
        label="Worst Cys3 backbone RMSD",
    )

    ax.scatter(
        plot_df["rank"],
        plot_df["max_cys3_sg_rmsd"],
        s=35,
        alpha=0.85,
        label="Worst Cys3 SG RMSD",
    )

    ax.axhline(
        CYS3_BACKBONE_RMSD_CUT,
        linestyle="--",
        linewidth=1,
        label=f"Backbone cutoff = {CYS3_BACKBONE_RMSD_CUT:.2f} Å",
    )

    ax.axhline(
        CYS3_SG_RMSD_CUT,
        linestyle=":",
        linewidth=1,
        label=f"SG cutoff = {CYS3_SG_RMSD_CUT:.2f} Å",
    )

    ax.set_xlabel("RF3 design rank")
    ax.set_ylabel("Worst Cys3 motif RMSD (Å)")
    ax.set_title("RF3 designs ranked by Pb-binding Cys3 motif preservation")
    ax.legend()
    fig.tight_layout()

    fig.savefig(OUT_PLOT, dpi=300)
    plt.show()

    print("Saved plot:", OUT_PLOT)
````

````python
def flatten_scalar_dict(d, prefix=""):
    out = {}
    if not isinstance(d, dict):
        return out
    for k, v in d.items():
        key = f"{prefix}.{k}" if prefix else str(k)
        if isinstance(v, dict):
            out.update(flatten_scalar_dict(v, key))
        elif isinstance(v, (str, int, float, bool)) or v is None:
            out[key] = v
        elif isinstance(v, list) and len(v) <= 20:
            out[key] = str(v)
    return out


def read_score_or_json(path):
    text = Path(path).read_text().strip()
    try:
        return json.loads(text)
    except Exception:
        pass
    data = {}
    for line in text.splitlines():
        line = line.strip()
        if not line or line.startswith("#"):
            continue
        if ":" in line:
            k, v = line.split(":", 1)
        else:
            parts = line.split(None, 1)
            if len(parts) != 2:
                continue
            k, v = parts
        k = k.strip(); v = v.strip()
        try:
            v = float(v)
        except ValueError:
            pass
        data[k] = v
    return data


OUT_CSV = scores_dir / "rf3_gathered_scores.csv"
rows = []
score_files = sorted(Path(rf3_out_dir).glob("**/*.score"))
json_files = sorted(Path(rf3_out_dir).glob("**/*.json"))

print("Score files:", len(score_files))
print("JSON files:", len(json_files))

for p in score_files:
    try:
        data = read_score_or_json(p)
    except Exception as exc:
        print(f"[WARNING] Could not read {p}: {exc}")
        continue
    row = flatten_scalar_dict(data)
    row["score_file"] = str(p)
    row["design_id"] = p.parent.name
    row["model_idx"] = int(re.search(r"model_(\d+)", p.stem).group(1)) if re.search(r"model_(\d+)", p.stem) else -1
    row["rf3_structure_path"] = str(p.with_suffix(".pdb")) if p.with_suffix(".pdb").exists() else ""
    rows.append(row)

if not rows:
    print("No RF3 results yet. Submit RF3 first.")
else:
    df_rf3_scores = pd.DataFrame(rows)
    df_rf3_scores.to_csv(OUT_CSV, index=False)
    print(f"Wrote {len(df_rf3_scores)} rows -> {OUT_CSV}")
    display(df_rf3_scores.head())
````

## 7. Score and rank designs

RMSD is computed between the RF3 output (scaffold chain A) and both Cys3 motif copies
from the preparation-stage double-motif PDB.

Both copies are evaluated independently. A design passes if **both** copies preserve the Cys3 triad.


````python
from pathlib import Path
import json
import re
import math

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import biotite.structure as struc
from biotite.structure.io import load_structure as bio_load


# ============================================================
# 0. Paths and settings
# ============================================================

PROJECT_ROOT = Path("/dtu/blackhole/00/222976/lead_sensoring")

EXP_DESIGN_DIR = PROJECT_ROOT / "work/exp_doubles_01"
EXP_RUN_DIR = PROJECT_ROOT / "work/exp_doubles_01_rfd3_mpnn_rf3"

DOUBLE_MOTIF_REFERENCE_DIR = EXP_DESIGN_DIR / "double_motif_structures"

# This should point to the folder containing RFD3 diffusion output JSONs.
# If your variable `diffusion_out_dir` already exists and is correct, this will use it.
try:
    DIFFUSION_OUT_DIR = Path(diffusion_out_dir)
except NameError:
    DIFFUSION_OUT_DIR = EXP_RUN_DIR / "diffusion_out"

SCORES_DIR = EXP_RUN_DIR / "scores"
PLOTS_DIR = EXP_RUN_DIR / "plots"

SCORES_DIR.mkdir(parents=True, exist_ok=True)
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

OUT_RMSD = SCORES_DIR / "rf3_double_motif_vs_original_motif_rmsd.csv"

# Pb-binding Cys3 only.
MOTIF1_CYS3_KEYS = ["C78", "D113", "D122"]
MOTIF2_CYS3_KEYS = ["X78", "Y113", "Y122"]

BACKBONE_ATOMS = ["N", "CA", "C", "O"]
SG_ATOMS = ["SG"]

PTM_CUT = 0.80
CYS3_BACKBONE_RMSD_CUT = 1.00
CYS3_SG_RMSD_CUT = 1.00

STRUCTURE_SUFFIXES = (".cif", ".pdb", ".cif.gz", ".pdb.gz")


# ============================================================
# 1. Basic structure helpers
# ============================================================

def load_atom_array(path):
    """Load a PDB/mmCIF file as a Biotite AtomArray."""
    path = Path(path)

    if not path.is_file():
        raise FileNotFoundError(f"Structure file not found: {path}")

    if str(path).endswith(".gz"):
        import gzip
        tmp = Path("/tmp/_tmp_biotite_load")
        if str(path).endswith(".cif.gz"):
            tmp = tmp.with_suffix(".cif")
        elif str(path).endswith(".pdb.gz"):
            tmp = tmp.with_suffix(".pdb")
        else:
            raise ValueError(f"Unsupported compressed structure format: {path}")

        with gzip.open(path, "rt") as f:
            tmp.write_text(f.read())

        arr = bio_load(str(tmp))
    else:
        arr = bio_load(str(path))

    # Some readers return AtomArrayStack. Use first model.
    if isinstance(arr, struc.AtomArrayStack):
        arr = arr[0]

    return arr


def parse_chain_res(label):
    """Parse residue labels like A42, C78, X113."""
    m = re.match(r"^([A-Za-z]+)(\d+)([A-Za-z]?)$", str(label))
    if not m:
        raise ValueError(f"Cannot parse residue label: {label}")
    return m.group(1), int(m.group(2))


def get_atom_coord(arr, chain, res_id, atom_name):
    """Return one atom coordinate from an AtomArray."""
    mask = (
        (arr.chain_id.astype(str) == str(chain))
        & (arr.res_id == int(res_id))
        & (arr.atom_name.astype(str) == str(atom_name))
    )

    if mask.sum() == 0:
        raise KeyError(f"{chain}{res_id}@{atom_name} not found")

    return arr.coord[mask][0]


def resolve_chain(atom_array, preferred_chain):
    """
    Use preferred_chain if present.

    If the structure has only one chain, use that single chain.
    This handles RF3 outputs that collapse everything to chain A.
    """
    chains = sorted(set(str(c) for c in atom_array.chain_id))

    if str(preferred_chain) in chains:
        return str(preferred_chain)

    if len(chains) == 1:
        return chains[0]

    raise KeyError(
        f"Chain {preferred_chain} not found. Available chains: {chains}"
    )


def kabsch_rmsd(P, Q):
    """Compute standard Kabsch RMSD between paired coordinates."""
    P = np.asarray(P, dtype=float)
    Q = np.asarray(Q, dtype=float)

    if P.shape != Q.shape:
        raise ValueError(f"Shape mismatch: P{P.shape} vs Q{Q.shape}")

    P_cent = P - P.mean(axis=0)
    Q_cent = Q - Q.mean(axis=0)

    H = P_cent.T @ Q_cent
    U, _, Vt = np.linalg.svd(H)

    d = np.linalg.det(Vt.T @ U.T)
    D = np.diag([1.0, 1.0, d])

    R = Vt.T @ D @ U.T
    P_rot = P_cent @ R.T

    return float(np.sqrt(np.mean(np.sum((P_rot - Q_cent) ** 2, axis=1))))


def rmsd_to_score(x, scale):
    """Convert RMSD to a soft score in [0, 1]."""
    if not np.isfinite(x):
        return np.nan
    return float(np.exp(-0.5 * (float(x) / float(scale)) ** 2))


# ============================================================
# 2. ID helpers
# ============================================================

def strip_all_suffixes(name):
    """Remove common file and RF3 suffixes."""
    s = Path(str(name)).name

    for suffix in [".cif.gz", ".pdb.gz", ".cif", ".pdb", ".json", ".csv"]:
        if s.endswith(suffix):
            s = s[: -len(suffix)]

    s = re.sub(r"__b\d+_d\d+$", "", s)
    s = re.sub(r"_model_\d+$", "", s)

    return s


def ori_tag_from_text(text):
    """Extract ori13A, ori15A, ori18A style tag."""
    m = re.search(r"ori\d+A", str(text))
    return m.group(0) if m else None


def clean_motif_id_from_text(text):
    """Extract clean 5gpe double-motif ID from a longer RF3/RFD3 name."""
    m = re.search(
        r"(5gpe_pb_motif_r\d+_ori\d+A_single_double_ori\d+A_rot\d+deg(?:_\d+)?)",
        str(text),
    )
    return m.group(1) if m else None


def rf3_source_id_from_design_id(design_id):
    """Recover the RF3 source ID by removing model/batch suffixes."""
    return strip_all_suffixes(design_id)


def make_rfd3_aliases(primary_id, condition=""):
    """Generate aliases that may refer to the same RFD3 design."""
    aliases = set()

    primary_id = str(primary_id)
    condition = str(condition)

    aliases.add(primary_id)
    aliases.add(strip_all_suffixes(primary_id))

    clean_id = clean_motif_id_from_text(primary_id)
    if clean_id:
        aliases.add(clean_id)
        aliases.add(re.sub(r"_\d+$", "", clean_id))

    ori_tag = ori_tag_from_text(primary_id)
    if ori_tag:
        rf3_prefix = f"rfd3_input_{ori_tag}_single_double_{ori_tag}_rot180deg"
        aliases.add(f"{rf3_prefix}_{primary_id}")

        if clean_id:
            aliases.add(f"{rf3_prefix}_{clean_id}")
            aliases.add(f"{rf3_prefix}_{re.sub(r'_\\d+$', '', clean_id)}")

    if condition:
        aliases.add(f"{condition}_{primary_id}")

    return {a for a in aliases if a}


def get_reference_path_for_design(source_id, rec):
    """Find the original double-motif reference PDB for this design."""
    pdb_files = sorted(DOUBLE_MOTIF_REFERENCE_DIR.glob("*.pdb"))

    if not pdb_files:
        raise FileNotFoundError(
            f"No reference PDB files found in {DOUBLE_MOTIF_REFERENCE_DIR}"
        )

    search_text = " ".join(
        str(x)
        for x in [
            source_id,
            rec.get("primary_id", ""),
            rec.get("condition", ""),
            rec.get("json_path", ""),
        ]
    )

    ori_tag = ori_tag_from_text(search_text)

    if ori_tag is None:
        raise ValueError(f"Could not infer ori tag from: {search_text}")

    matches = [p for p in pdb_files if ori_tag in p.stem]

    if len(matches) != 1:
        raise ValueError(
            f"Expected one reference for {ori_tag}, found {len(matches)}: "
            f"{[p.name for p in matches]}"
        )

    return matches[0]


# ============================================================
# 3. Build RFD3 mapping records
# ============================================================

def build_rfd3_records(diffusion_out_dir):
    """Build RFD3 mapping records with robust aliases."""
    diffusion_out_dir = Path(diffusion_out_dir)

    json_files = sorted(diffusion_out_dir.glob("*/*.json"))

    if not json_files:
        raise FileNotFoundError(
            f"No RFD3 JSON files found under {diffusion_out_dir}/*/*.json"
        )

    primary_records = {}
    alias_records = {}

    for jp in json_files:
        with jp.open() as f:
            data = json.load(f)

        mapping = data.get("diffused_index_map", {})

        if not mapping:
            continue

        primary_id = jp.stem
        condition = jp.parent.name

        rec = {
            "primary_id": primary_id,
            "condition": condition,
            "mapping": mapping,
            "json_path": str(jp),
        }

        primary_records[primary_id] = rec

        aliases = make_rfd3_aliases(primary_id, condition)
        rec["aliases"] = sorted(aliases)

        for alias in aliases:
            if alias in alias_records and alias_records[alias]["json_path"] != rec["json_path"]:
                print("[WARNING] Skipping ambiguous RFD3 alias:", alias)
                continue
            alias_records[alias] = rec

    print("RFD3 primary records:", len(primary_records))
    print("RFD3 lookup aliases:", len(alias_records))

    return primary_records, alias_records


def resolve_rfd3_record(source_id, rfd3_alias_records):
    """Resolve an RF3 source ID to an RFD3 record."""
    source_id = str(source_id)

    candidates = set()
    candidates.add(source_id)
    candidates.add(strip_all_suffixes(source_id))

    clean_id = clean_motif_id_from_text(source_id)
    if clean_id:
        candidates.add(clean_id)
        candidates.add(re.sub(r"_\d+$", "", clean_id))

    for candidate in candidates:
        if candidate in rfd3_alias_records:
            return candidate, rfd3_alias_records[candidate]

    raise KeyError(
        "Could not resolve RFD3 record.\n"
        f"source_id: {source_id}\n"
        f"candidates: {sorted(candidates)}\n"
        f"first aliases: {list(rfd3_alias_records.keys())[:10]}"
    )


rfd3_primary_records, rfd3_alias_records = build_rfd3_records(DIFFUSION_OUT_DIR)


# ============================================================
# 4. Collect RF3 structures and optional pTM values
# ============================================================

def find_rf3_structure_files(run_dir):
    """Find candidate RF3 output structures."""
    run_dir = Path(run_dir)

    files = [
        p for p in run_dir.rglob("*")
        if p.is_file() and str(p).endswith(STRUCTURE_SUFFIXES)
    ]

    # Avoid obvious non-RF3 references if they ever appear under the run dir.
    bad_tokens = [
        "double_motif_structures",
        "motif_scan",
        "rfdiffusion_inputs",
    ]

    files = [
        p for p in files
        if not any(tok in str(p) for tok in bad_tokens)
    ]

    return sorted(files)


def infer_design_id_from_structure_path(path):
    """Infer a design ID from an RF3 output structure path."""
    path = Path(path)
    stem = strip_all_suffixes(path.name)

    # If file name is generic, use parent folder.
    generic_names = {
        "model",
        "model_0",
        "prediction",
        "structure",
        "output",
        "fold",
        "rank_0",
    }

    if stem in generic_names or len(stem) < 12:
        stem = strip_all_suffixes(path.parent.name)

    return stem


def recursive_find_key(obj, wanted_keys):
    """Recursively find numeric values for selected keys in JSON-like data."""
    hits = []

    if isinstance(obj, dict):
        for k, v in obj.items():
            if str(k).lower() in wanted_keys:
                try:
                    hits.append(float(v))
                except Exception:
                    pass
            hits.extend(recursive_find_key(v, wanted_keys))

    elif isinstance(obj, list):
        for x in obj:
            hits.extend(recursive_find_key(x, wanted_keys))

    return hits


def find_ptm_near_structure(structure_path):
    """Try to find pTM in JSON files close to the RF3 structure."""
    structure_path = Path(structure_path)

    candidate_jsons = []
    candidate_jsons.extend(structure_path.parent.glob("*.json"))
    candidate_jsons.extend(structure_path.parent.parent.glob("*.json"))

    wanted = {"ptm", "predicted_tm_score", "tm_score"}

    for jp in sorted(set(candidate_jsons)):
        try:
            with jp.open() as f:
                data = json.load(f)

            hits = recursive_find_key(data, wanted)
            if hits:
                return float(hits[0]), str(jp)

        except Exception:
            continue

    return np.nan, ""


def build_rf3_output_table(run_dir):
    """Build a clean RF3 output table from structures on disk."""
    structure_files = find_rf3_structure_files(run_dir)

    if not structure_files:
        raise FileNotFoundError(f"No RF3 structure files found under {run_dir}")

    rows = []

    for p in structure_files:
        design_id = infer_design_id_from_structure_path(p)
        ptm, ptm_source = find_ptm_near_structure(p)

        rows.append(
            {
                "design_id": design_id,
                "source_id_raw": rf3_source_id_from_design_id(design_id),
                "rf3_structure_path": str(p),
                "ptm": ptm,
                "ptm_source": ptm_source,
            }
        )

    df = pd.DataFrame(rows)

    # Remove duplicate structures if exact same path appears.
    df = df.drop_duplicates(subset=["rf3_structure_path"]).reset_index(drop=True)

    print("RF3 structures found:", len(df))
    display(df.head())

    return df


df_rf3_outputs = build_rf3_output_table(EXP_RUN_DIR)


# ============================================================
# 5. Motif RMSD calculation
# ============================================================

def collect_motif_rmsd(rf3_arr, ref_arr, mapping, motif_keys, atoms):
    """
    Compare RF3 final motif to the original double-motif reference.

    RF3 side:
        mapping[key], e.g. C78 -> A48

    Reference side:
        key itself, e.g. C78 in the original double-motif PDB.
    """
    P = []
    Q = []
    used_atoms = []
    missing_atoms = []

    for key in motif_keys:
        if key not in mapping:
            missing_atoms.append(f"{key}:missing_from_mapping")
            continue

        rf3_label = str(mapping[key])
        rf3_chain_raw, rf3_res = parse_chain_res(rf3_label)

        ref_chain_raw, ref_res = parse_chain_res(key)

        try:
            rf3_chain = resolve_chain(rf3_arr, rf3_chain_raw)
            ref_chain = resolve_chain(ref_arr, ref_chain_raw)
        except Exception as exc:
            missing_atoms.append(f"{key}->{rf3_label}:chain_error:{exc}")
            continue

        for atom in atoms:
            try:
                rf3_coord = get_atom_coord(rf3_arr, rf3_chain, rf3_res, atom)
                ref_coord = get_atom_coord(ref_arr, ref_chain, ref_res, atom)

                P.append(rf3_coord)
                Q.append(ref_coord)
                used_atoms.append(f"{key}->{rf3_label}@{atom}")

            except Exception:
                missing_atoms.append(f"{key}->{rf3_label}@{atom}")

    if len(P) < 3:
        return np.nan, len(P), ",".join(used_atoms), ",".join(missing_atoms)

    return (
        kabsch_rmsd(np.asarray(P), np.asarray(Q)),
        len(P),
        ",".join(used_atoms),
        ",".join(missing_atoms),
    )


def check_cys_preserved(rf3_arr, mapping, motif_keys):
    """Check whether RF3 kept the mapped motif residues as CYS."""
    checked = []
    failed = []

    for key in motif_keys:
        if key not in mapping:
            failed.append(f"{key}:missing_from_mapping")
            return False, ",".join(checked), ",".join(failed)

        rf3_label = str(mapping[key])
        chain_raw, res_id = parse_chain_res(rf3_label)

        try:
            chain = resolve_chain(rf3_arr, chain_raw)
        except Exception as exc:
            failed.append(f"{key}->{rf3_label}:chain_error:{exc}")
            return False, ",".join(checked), ",".join(failed)

        mask = (
            (rf3_arr.chain_id.astype(str) == str(chain))
            & (rf3_arr.res_id == int(res_id))
        )

        if mask.sum() == 0:
            failed.append(f"{key}->{rf3_label}:residue_not_found")
            return False, ",".join(checked), ",".join(failed)

        res_name = str(rf3_arr.res_name[mask][0])
        checked.append(f"{key}->{rf3_label}:{res_name}")

        if res_name != "CYS":
            failed.append(f"{key}->{rf3_label}:expected_CYS_found_{res_name}")
            return False, ",".join(checked), ",".join(failed)

    return True, ",".join(checked), ",".join(failed)


rmsd_rows = []

for _, row in df_rf3_outputs.iterrows():
    design_id = row["design_id"]
    source_id_raw = row["source_id_raw"]
    rf3_path = Path(row["rf3_structure_path"])

    try:
        resolved_alias, rec = resolve_rfd3_record(source_id_raw, rfd3_alias_records)
        mapping = rec["mapping"]

        ref_path = get_reference_path_for_design(source_id_raw, rec)

        rf3_arr = load_atom_array(rf3_path)
        ref_arr = load_atom_array(ref_path)

        kept1, cys_checked1, cys_failed1 = check_cys_preserved(
            rf3_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
        )

        kept2, cys_checked2, cys_failed2 = check_cys_preserved(
            rf3_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
        )

        bb1, bb_n1, bb_atoms1, bb_missing1 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
            BACKBONE_ATOMS,
        )

        bb2, bb_n2, bb_atoms2, bb_missing2 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
            BACKBONE_ATOMS,
        )

        sg1, sg_n1, sg_atoms1, sg_missing1 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF1_CYS3_KEYS,
            SG_ATOMS,
        )

        sg2, sg_n2, sg_atoms2, sg_missing2 = collect_motif_rmsd(
            rf3_arr,
            ref_arr,
            mapping,
            MOTIF2_CYS3_KEYS,
            SG_ATOMS,
        )

        ptm = float(row.get("ptm", np.nan))

        max_bb_rmsd = np.nanmax([bb1, bb2])
        max_sg_rmsd = np.nanmax([sg1, sg2])

        passes_first_triage = (
            kept1
            and kept2
            and np.isfinite(bb1)
            and np.isfinite(bb2)
            and np.isfinite(sg1)
            and np.isfinite(sg2)
            and bb1 <= CYS3_BACKBONE_RMSD_CUT
            and bb2 <= CYS3_BACKBONE_RMSD_CUT
            and sg1 <= CYS3_SG_RMSD_CUT
            and sg2 <= CYS3_SG_RMSD_CUT
            and (not np.isfinite(ptm) or ptm >= PTM_CUT)
        )

        rmsd_rows.append(
            {
                "design_id": design_id,
                "source_id_raw": source_id_raw,
                "resolved_rfd3_alias": resolved_alias,
                "rfd3_primary_id": rec["primary_id"],
                "condition": rec["condition"],

                "rf3_structure_path": str(rf3_path),
                "original_motif_reference_path": str(ref_path),
                "rfd3_json_path": rec["json_path"],

                "ptm": ptm,
                "ptm_source": row.get("ptm_source", ""),

                "kept_cys3_motif1": kept1,
                "kept_cys3_motif2": kept2,
                "kept_both_cys3": kept1 and kept2,

                "cys_checked_motif1": cys_checked1,
                "cys_checked_motif2": cys_checked2,
                "cys_failed_motif1": cys_failed1,
                "cys_failed_motif2": cys_failed2,

                "cys3_backbone_rmsd_motif1": bb1,
                "cys3_backbone_rmsd_motif2": bb2,
                "cys3_sg_rmsd_motif1": sg1,
                "cys3_sg_rmsd_motif2": sg2,

                "max_cys3_backbone_rmsd": max_bb_rmsd,
                "max_cys3_sg_rmsd": max_sg_rmsd,

                "n_backbone_atoms_motif1": bb_n1,
                "n_backbone_atoms_motif2": bb_n2,
                "n_sg_atoms_motif1": sg_n1,
                "n_sg_atoms_motif2": sg_n2,

                "missing_backbone_atoms_motif1": bb_missing1,
                "missing_backbone_atoms_motif2": bb_missing2,
                "missing_sg_atoms_motif1": sg_missing1,
                "missing_sg_atoms_motif2": sg_missing2,

                "cys3_backbone_score_motif1": rmsd_to_score(
                    bb1,
                    CYS3_BACKBONE_RMSD_CUT,
                ),
                "cys3_backbone_score_motif2": rmsd_to_score(
                    bb2,
                    CYS3_BACKBONE_RMSD_CUT,
                ),
                "cys3_sg_score_motif1": rmsd_to_score(
                    sg1,
                    CYS3_SG_RMSD_CUT,
                ),
                "cys3_sg_score_motif2": rmsd_to_score(
                    sg2,
                    CYS3_SG_RMSD_CUT,
                ),

                "passes_first_triage": passes_first_triage,
            }
        )

    except Exception as exc:
        print("[FAILED]", design_id, exc)


df_rmsd = pd.DataFrame(rmsd_rows)
df_rmsd.to_csv(OUT_RMSD, index=False)

print("Evaluated RF3 structures:", len(df_rmsd))
print("Passing first triage:", int(df_rmsd["passes_first_triage"].sum()) if len(df_rmsd) else 0)
print("Saved:", OUT_RMSD)

display(df_rmsd.head())
````

````python
from pathlib import Path
import shutil
import pandas as pd
import numpy as np
import re

# ============================================================
# Select and export the best TRUE RF3 design
# Only designs present in rf3_manifest.csv are allowed.
# ============================================================

RUN_DIR = Path("/dtu/blackhole/00/222976/lead_sensoring/work/exp_doubles_01_rfd3_mpnn_rf3")
SCORES_DIR = RUN_DIR / "scores"
CONFIGS_DIR = RUN_DIR / "configs"
EXPORT_DIR = RUN_DIR / "best_true_rf3_export"

RANKED_CSV = SCORES_DIR / "rf3_double_motif_ranked_by_motif_rmsd.csv"
RMSD_CSV = SCORES_DIR / "rf3_double_motif_vs_original_motif_rmsd.csv"
RF3_MANIFEST = CONFIGS_DIR / "rf3_manifest.csv"

EXPORT_DIR.mkdir(parents=True, exist_ok=True)

if not RF3_MANIFEST.is_file():
    raise FileNotFoundError(f"RF3 manifest not found: {RF3_MANIFEST}")

score_csv = RANKED_CSV if RANKED_CSV.is_file() else RMSD_CSV

if not score_csv.is_file():
    raise FileNotFoundError(f"No score table found: {RANKED_CSV} or {RMSD_CSV}")

df_scores = pd.read_csv(score_csv)
df_manifest = pd.read_csv(RF3_MANIFEST)

if df_scores.empty:
    raise SystemExit("Score table is empty.")

if "name" not in df_manifest.columns:
    raise ValueError("rf3_manifest.csv must contain a 'name' column.")

if "design_id" not in df_scores.columns:
    raise ValueError("Score table must contain a 'design_id' column.")


def remove_repeated_best_prefix(x: str) -> str:
    """Remove repeated export prefixes from an identifier."""
    x = str(x)
    while x.startswith("BEST_RF3_"):
        x = x[len("BEST_RF3_"):]
    return x


def strip_structure_suffixes(filename: str) -> str:
    """Remove common structure suffixes from a filename."""
    x = str(filename)
    for suffix in [".cif.gz", ".pdb.gz", ".cif", ".pdb"]:
        if x.endswith(suffix):
            return x[: -len(suffix)]
    return Path(x).stem


def normalize_design_id(x: str) -> str:
    """Normalize a design identifier for matching."""
    x = remove_repeated_best_prefix(str(x))
    x = strip_structure_suffixes(x)
    return x


def normalize_path_to_id(path_like: str) -> str:
    """Convert a structure path to a normalized design identifier."""
    if pd.isna(path_like):
        return ""
    return normalize_design_id(Path(str(path_like)).name)


# Normalize score identifiers.
df_scores["_design_id_norm"] = df_scores["design_id"].map(normalize_design_id)

if "rf3_structure_path" in df_scores.columns:
    df_scores["_path_id_norm"] = df_scores["rf3_structure_path"].map(normalize_path_to_id)
else:
    df_scores["_path_id_norm"] = ""

# Normalize manifest identifiers.
df_manifest["_manifest_name_norm"] = df_manifest["name"].map(normalize_design_id)

# First, require exact match to manifest name.
manifest_names = set(df_manifest["_manifest_name_norm"])

df_scores["_matches_manifest_by_design_id"] = df_scores["_design_id_norm"].isin(manifest_names)
df_scores["_matches_manifest_by_path"] = df_scores["_path_id_norm"].isin(manifest_names)

df_true_rf3 = df_scores[
    df_scores["_matches_manifest_by_design_id"] | df_scores["_matches_manifest_by_path"]
].copy()

if df_true_rf3.empty:
    print("No exact RF3-manifest matches found in the score table.")
    print("This means your current score table is probably scoring/exporting diffusion backbones, not RF3 outputs.")
    print("\nManifest RF3 names:")
    print(df_manifest["name"].to_string(index=False))
    raise SystemExit("Recompute RMSD scores using only RF3 output structures from manifest names.")

# Choose the matching key for joining.
df_true_rf3["_join_name"] = np.where(
    df_true_rf3["_matches_manifest_by_design_id"],
    df_true_rf3["_design_id_norm"],
    df_true_rf3["_path_id_norm"],
)

df_joined = df_true_rf3.merge(
    df_manifest,
    left_on="_join_name",
    right_on="_manifest_name_norm",
    how="left",
    suffixes=("", "_manifest"),
)

# Convert numeric score columns.
numeric_cols = [
    "ptm",
    "max_cys3_sg_rmsd",
    "max_cys3_backbone_rmsd",
    "cys3_sg_rmsd_motif1",
    "cys3_sg_rmsd_motif2",
    "cys3_backbone_rmsd_motif1",
    "cys3_backbone_rmsd_motif2",
]

for col in numeric_cols:
    if col in df_joined.columns:
        df_joined[col] = pd.to_numeric(df_joined[col], errors="coerce")

if "passes_first_triage" in df_joined.columns:
    df_joined["passes_first_triage"] = (
        df_joined["passes_first_triage"]
        .astype(str)
        .str.lower()
        .isin(["true", "1", "yes"])
    )
else:
    df_joined["passes_first_triage"] = False

if "ptm" not in df_joined.columns:
    df_joined["ptm"] = np.nan

df_joined["_ptm_sort"] = df_joined["ptm"].fillna(-1)

required_sort_cols = [
    "passes_first_triage",
    "max_cys3_sg_rmsd",
    "max_cys3_backbone_rmsd",
    "_ptm_sort",
]

missing_sort_cols = [col for col in required_sort_cols if col not in df_joined.columns]
if missing_sort_cols:
    raise ValueError(f"Missing required ranking columns: {missing_sort_cols}")

df_ranked_true_rf3 = df_joined.sort_values(
    by=[
        "passes_first_triage",
        "max_cys3_sg_rmsd",
        "max_cys3_backbone_rmsd",
        "_ptm_sort",
    ],
    ascending=[
        False,
        True,
        True,
        False,
    ],
).reset_index(drop=True)

best = df_ranked_true_rf3.iloc[0]

best_rf3_name = best["_join_name"]
best_mpnn_fasta = Path(best["mpnn_fasta"])
best_rf3_json = Path(best["rf3_json"])

# Prefer structure path from score table, but reject exported copies.
if "rf3_structure_path" in best.index and str(best.get("rf3_structure_path", "")).strip():
    best_structure_path = Path(str(best["rf3_structure_path"]))
else:
    best_structure_path = None

if best_structure_path is None or not best_structure_path.is_file() or "best_design_export" in str(best_structure_path):
    candidates = []
    for pattern in [
        f"{best_rf3_name}*.cif.gz",
        f"{best_rf3_name}*.cif",
        f"{best_rf3_name}*.pdb.gz",
        f"{best_rf3_name}*.pdb",
    ]:
        candidates.extend(RUN_DIR.rglob(pattern))

    candidates = [
        p for p in candidates
        if "best_design_export" not in str(p)
        and "best_true_rf3_export" not in str(p)
    ]

    if not candidates:
        raise FileNotFoundError(
            f"Could not find original RF3 structure for: {best_rf3_name}"
        )

    best_structure_path = sorted(candidates)[0]

# Export cleanly.
suffix = "".join(best_structure_path.suffixes)
best_structure_export = EXPORT_DIR / f"BEST_TRUE_RF3_{best_rf3_name}{suffix}"
best_summary_export = EXPORT_DIR / "best_true_rf3_summary.csv"
top10_export = EXPORT_DIR / "top10_true_rf3_designs.csv"
manifest_row_export = EXPORT_DIR / "best_true_rf3_manifest_row.csv"

shutil.copy2(best_structure_path, best_structure_export)

pd.DataFrame([best.drop(labels=["_ptm_sort"], errors="ignore")]).to_csv(
    best_summary_export,
    index=False,
)

df_ranked_true_rf3.drop(columns=["_ptm_sort"], errors="ignore").head(10).to_csv(
    top10_export,
    index=False,
)

pd.DataFrame([{
    "rf3_name": best_rf3_name,
    "rf3_json": str(best_rf3_json),
    "mpnn_fasta": str(best_mpnn_fasta),
    "rf3_structure_original": str(best_structure_path),
    "rf3_structure_exported": str(best_structure_export),
}]).to_csv(
    manifest_row_export,
    index=False,
)

zip_base = EXPORT_DIR.parent / "best_true_rf3_export"
zip_path = shutil.make_archive(str(zip_base), "zip", EXPORT_DIR)

print("Best TRUE RF3 design selected:")
print("  RF3 name:", best_rf3_name)
print("  condition:", best.get("condition", ""))
print("  passes_first_triage:", best.get("passes_first_triage", ""))

print("\nScores:")
print("  pTM:", best.get("ptm", np.nan))
print("  max SG RMSD:", best.get("max_cys3_sg_rmsd", np.nan))
print("  max backbone RMSD:", best.get("max_cys3_backbone_rmsd", np.nan))
print("  motif1 SG RMSD:", best.get("cys3_sg_rmsd_motif1", np.nan))
print("  motif2 SG RMSD:", best.get("cys3_sg_rmsd_motif2", np.nan))
print("  motif1 backbone RMSD:", best.get("cys3_backbone_rmsd_motif1", np.nan))
print("  motif2 backbone RMSD:", best.get("cys3_backbone_rmsd_motif2", np.nan))

print("\nProvenance:")
print("  RF3 manifest:", RF3_MANIFEST)
print("  RF3 input JSON:", best_rf3_json)
print("  MPNN FASTA:", best_mpnn_fasta)
print("  Original RF3 structure:", best_structure_path)

print("\nExported files:")
print("  Best TRUE RF3 structure:", best_structure_export)
print("  Best summary:", best_summary_export)
print("  Top 10 TRUE RF3 table:", top10_export)
print("  Manifest row:", manifest_row_export)
print("  Zip:", zip_path)

print("\nTo download from your local terminal:")
print(f"  scp s252608@login1.hpc.dtu.dk:{zip_path} .")

print("\nTo inspect directly in ChimeraX:")
print(f"  open {best_structure_export}")
````
